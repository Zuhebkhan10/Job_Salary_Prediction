```python
import pandas as pd

df=pd.read_csv('dataset/job_salary.csv')

df.head()
```

```python
# . Selecting Specific Columns
# How would you select just the job_title and Salary columns?

res=df[['job_title','salary']]
res.head()
```

```python
# Filtering Rows:
# Can you show me all employees earning more than 90,000

res=df[df['salary']>90000]
res.head()

# Explanation: You create a boolean mask by writing a condition, then use it to filter the DataFrame.
# This is called boolean indexing.
```

```python
# Adding a New Column
# How would you add a Bonus column that's 20% of the Salary?

df['Bonus']=df['salary']*0.20
df.tail()

# Explanation: You can create new columns by simply assigning values to a new column name.
# The calculation is applied element-wise.
```

```python
# Sorting Data
# Could you sort the employees by salary from highest to lowest?

sorted_data=df.sort_values('salary',ascending=False)
sorted_data.head()
```

```python
# Handling Missing Values:
 # What's a good way to fill missing values with the column average?

filled=df.fillna(df.mean())
filled.head()

# Explanation: fillna() replaces missing values. Using mean() is a common strategy for numerical data,
# though the best approach depends on your specific use case.
```

```python
# Group By and Calculate Average
# How would you find the average salary for each department?

df.groupby('job_title')['salary'].mean()

# Explanation: groupby() is incredibly powerful for aggregating data. You group by a category, then apply an aggregation
# function like mean().
```

```python
#Multiple Aggregations at Once
# Can you get the count, average, and maximum salary by department?

df.groupby('job_title')['salary'].agg(['count','mean','max'])

# Explanation: The agg() method lets you apply multiple aggregation functions simultaneously, which is very efficient!
```

```python
# . Applying Custom Functions
# How would you categorize salaries as Low, Medium, or High?

def categorize_salary(salary):
    if salary<90000:
        return 'Low'
    elif salary<100000:
        return 'Medium'
    else:
        return 'High'
df['Category']=df['salary'].apply(categorize_salary)
df.head(10)

# Explanation: apply() lets you run custom functions on each element. It's very flexible for complex transformations.
```

```python
# . Working with Strings
# Question: Can you extract just the first name from a full name job_title?

df['FirstName']=df['job_title'].str.split().str[0]
df

df['LastName']=df['job_title'].str.split().str[1]
df.head()

# Explanation: The .str accessor gives you string methods. Here we split on spaces and take the first element.
```

```python
# Removing Duplicates
# Question: How do you remove duplicate rows from a DataFrame?

remove=df['industry'].drop_duplicates()
remove.head()

# Explanation: drop_duplicates() removes rows that are completely identical. You can also specify certain columns to
# check for duplicates.
```

```python
# Multiple Conditions in Filtering
# Question: How would you find job_tile earning more than 260,000?

res=df[(df['industry']=="Healthcare")& (df['salary']>260000)]
res.head()

# Explanation: rename() takes a dictionary mapping old names to new names. It's cleaner than reassigning the entire columns list.
```

```python
#  Renaming Columns
# How would you rename the Salary column to Annual_Salary?

res=df.rename(columns={'salary':'Annual_salary'})
res.head()

# Explanation: rename() takes a dictionary mapping old names to new names.
# It's cleaner than reassigning the entire columns list.
```

```python
# Managing Index
# How do you set the Name column as the index?

df_index=df.set_index('company_size')
print(df_index.head())

df_reset=df_index.reset_index()
df_reset.head()
```
